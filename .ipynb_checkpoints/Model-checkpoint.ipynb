# define problem statement
- identify the customers who are defalut and no default by previuous bill payments and customer demographic
- Target variable - default
- Classification model

```python
# target and input variables
```

```python
y=df['Default']
X=df.drop(columns=['ID','SEX','EDUCATION', 'MARRIAGE','Default','AGE'])
```

```python
# feature selection
```

```python
# Mutual information for classification
from sklearn. feature_selection import mutual_info_classif
mutual_info=mutual_info_classif(X,y)
mutual_info=pd.Series(mutual_info)
mutual_info.index=X.columns
mutual_info.sort_values(ascending=False)
```

```python
# k best features
```

```python
# using F_regression test for regression
from sklearn. feature_selection import SelectKBest, f_classif

select_columns=SelectKBest(score_func=f_classif, k=15)
best_features=select_columns.fit(X, y)

# to display best feature names
selected_feature=X.columns [best_features.get_support ()]
print(selected_feature)
```

```python
# use k best features
```

```python
X=df[['LIMIT_BAL', 'PAY_0', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6',
       'BILL_AMT1', 'BILL_AMT2', 'PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3',
       'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']]
y=df['Default']
```

```python
X.head()
```

```python
# train test split
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

# data preprocess

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()

X_train_trans=scaler.fit_transform(X_train)
X_test_trans=scaler.transform(X_test)
```

# KNN Model

### hyper parameter tunning

### grid search

```python
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.metrics import classification_report
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
tuned_parameters = [{'n_neighbors': [i for i in range(1, 51)],'p': [1, 2, 3]}]

clf = GridSearchCV(estimator=KNeighborsClassifier(),
                    param_grid=tuned_parameters,
                    scoring='accuracy',
                    cv=5,
                    return_train_score=True,
                    verbose=1 ) # optional
clf.fit(X_train_trans, y_train)
```
