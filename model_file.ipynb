```python
df=df.rename(columns={'default ':'Default'})
```

```python
# to balance the classes

majority = df[df['Default'] == 'N']
minority = df[df['Default'] == 'Y']

majority_downsampled = majority.sample(
    n=14512,
    random_state=42
)

df = pd.concat([majority_downsampled, minority])
df = df.sample(frac=1, random_state=42)
```

# define problem statement
- identify the customers who are defalut and no default by previuous bill payments and customer demographic
- Target variable - default
- Classification model

```python
# target and input variables
```

```python
y=df['Default']
X=df.drop(columns=['ID','SEX','EDUCATION', 'MARRIAGE','Default','AGE'])
```

```python
# feature selection
```

```python
# Mutual information for classification
from sklearn. feature_selection import mutual_info_classif
mutual_info=mutual_info_classif(X,y)
mutual_info=pd.Series(mutual_info)
mutual_info.index=X.columns
mutual_info.sort_values(ascending=False)
```

# train test split

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,stratify=y)
```

# data preprocess

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()

X_train_trans=scaler.fit_transform(X_train)
X_test_trans=scaler.transform(X_test)
```

# base model

### knn model

```python
from sklearn.metrics import classification_report
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model=KNeighborsClassifier()
knn_model.fit(X_train_trans,y_train)
```

```python
# prediction
y_pred_knn=knn_model.predict(X_test_trans)
```

```python
# evaluaion
from sklearn.metrics import accuracy_score
from sklearn.metrics import f1_score,precision_score,recall_score

knn_accuracy=accuracy_score(y_test,y_pred_knn)
knn_f1 = f1_score(y_test, y_pred_knn, average='macro')
knn_recall = recall_score(y_test, y_pred_knn, average='macro')
knn_precision = precision_score(y_test, y_pred_knn, average='macro')
knn_accuracy
```

```python
# results
import pandas as pd

results={'Algorithm':['knn'],
         'accuracy':[knn_accuracy],
         'precision':[knn_precision],
         'recall':[knn_recall],
         'f1_score':[knn_f1]}
```

```python
results=pd.DataFrame(results)
```

```python
results
```

# classification report

```python
from sklearn.metrics import classification_report



knn_report=classification_report(y_test, y_pred_knn)
print(knn_report)
```
